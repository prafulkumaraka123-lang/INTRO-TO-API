## Q1. What is FastAPI and why is it used to build backend APIs?

**Answer:** FastAPI is a modern Python web framework for building backend APIs. It receives HTTP requests, processes them using application logic, and returns responses, commonly as JSON. It uses Python type hints and supports automatic API documentation.

A normal Python script generally executes its instructions and stops. A FastAPI application runs under a server such as Uvicorn and keeps listening for incoming requests. It can respond repeatedly while the server is running.

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello, FastAPI!"}

# Save as main.py and run: uvicorn main:app --reload
````

## Q2. Traditional web development vs FastAPI-based API development

**Answer:** Traditional web development commonly returns a complete HTML page that a browser displays. FastAPI-based API development commonly returns structured data, usually JSON, which can be used by a website, mobile app, or another service.

- Traditional web response: `<h1>Welcome</h1>` (HTML).
- FastAPI response: `{"message": "Welcome"}` (JSON).

````python
html_response = "<h1>Welcome</h1>"
api_response = {"message": "Welcome"}

print("Traditional web response:", html_response)
print("FastAPI-style API response:", api_response)
````

## Q4. What is Uvicorn and why is an ASGI server needed?

**Answer:** Uvicorn is a lightweight ASGI (Asynchronous Server Gateway Interface) server used to run FastAPI applications. FastAPI defines the routes and request-handling logic, while Uvicorn listens for network requests and connects clients to the application.

Request-response flow:
1. Uvicorn listens for HTTP requests.
2. A client sends a request.
3. Uvicorn passes it to FastAPI through ASGI.
4. FastAPI matches the route and executes its function.
5. FastAPI creates a response and Uvicorn sends it to the client.

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Request received by FastAPI"}

# Start with: uvicorn main:app --reload
````

## Q6. Explain the GET method

**Answer:** GET is an HTTP method used to retrieve data from a server. It is considered safe because its intended purpose is reading information, not changing server state. A correctly designed GET endpoint should not create, update, or delete records.

Examples in a student management system:
1. `GET /students/101` retrieves the details of student 101.
2. `GET /students` retrieves the list of students.

````python
from fastapi import FastAPI

app = FastAPI()

students = [
    {"id": 101, "name": "Asha", "course": "Electrical Engineering"},
    {"id": 102, "name": "Rahul", "course": "Computer Science"}
]

@app.get("/students")
def get_students():
    return students

@app.get("/students/{student_id}")
def get_student(student_id: int):
    for student in students:
        if student["id"] == student_id:
            return student
    return {"message": "Student not found"}
````

## Q7. Create `/`, `/about`, and `/courses` GET endpoints

**Answer:** The following `main.py` defines all three requested endpoints. `/` returns a welcome message, `/about` returns the app name and description, and `/courses` returns three course names.

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def welcome():
    return {"message": "Welcome to the FastAPI application!"}

@app.get("/about")
def about():
    return {
        "app_name": "Student Course API",
        "description": "An API that provides information about available courses."
    }

@app.get("/courses")
def get_courses():
    return {"courses": [
        "Electrical Engineering",
        "Computer Science",
        "Mechanical Engineering"
    ]}
````

## Q9. What is an API endpoint?

**Answer:** An API endpoint is a URL path, used with an HTTP method, where a client can request a resource or operation.

Examples:
- `/` – root or welcome endpoint.
- `/about` – information about the application.
- `/student` – may represent a student resource.
- `/students` – commonly represents a collection of students.

Clear endpoint names are short, descriptive, consistent, and identify the resource. For example, `/students` indicates a collection, while `/students/101` identifies one student. If a requested path and method do not match a defined route, FastAPI returns **404 Not Found**.

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Welcome"}

@app.get("/about")
def about():
    return {"app": "Student API"}

@app.get("/student")
def student():
    return {"id": 101, "name": "Asha"}

@app.get("/students")
def students():
    return [{"id": 101, "name": "Asha"}, {"id": 102, "name": "Rahul"}]

# An undefined path such as /teachers returns HTTP 404.
````

## Q10. Build a small Library API

**Answer:** This application provides the five required GET endpoints. It uses an in-memory list of dictionaries as sample book data.

````python
# Complete main.py code for the Library API
from fastapi import FastAPI

app = FastAPI(title="Library API")

books = [
    {"id": 1, "title": "The Alchemist", "author": "Paulo Coelho", "available": True},
    {"id": 2, "title": "Wings of Fire", "author": "A. P. J. Abdul Kalam", "available": True},
    {"id": 3, "title": "Atomic Habits", "author": "James Clear", "available": False}
]

@app.get("/")
def home():
    return {"message": "Welcome to the Library API"}

@app.get("/library-info")
def library_info():
    return {
        "library_name": "Central City Library",
        "total_books": len(books),
        "working_hours": "9:00 AM to 6:00 PM"
    }

@app.get("/books")
def get_books():
    return [
        {"id": b["id"], "title": b["title"], "author": b["author"]}
        for b in books
    ]

@app.get("/available-books")
def get_available_books():
    return [
        {"id": b["id"], "title": b["title"], "author": b["author"]}
        for b in books if b["available"]
    ]

@app.get("/featured-book")
def get_featured_book():
    return {
        "id": 2,
        "title": "Wings of Fire",
        "reason": "An inspiring autobiography about learning, science, and perseverance."
    }
````
